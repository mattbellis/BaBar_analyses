```python
def return_cart(particle='K_S'):
    
    r,ct,phi = d[f'{particle}p3'], d[f'{particle}costh'], d[f'{particle}phi']
    #print(r,ct,phi)
    cart = bat.spherical_to_cartesian(r,ct,phi)

    x,y,z = cart[0][:], cart[1][:], cart[2][:]
    
    #print(cart)
    #print(x, y, z)

    return x,y,z



def print_columns(x1,x2,x3):
    for a,b,c in zip(x1,x2,x3):
        print(f'{a:11.7f} {b:11.7f}  {c:11.7f}')
```
